# Repaso de trigonometría para ángulos entre vectores

Este cuaderno es un recurso de apoyo. No introduce números complejos. Su objetivo inmediato es ayudar a interpretar

$$\cos\theta=\frac{\langle x,y\rangle}{\|x\|\,\|y\|}. $$

Volveremos a enlazarlo en la Unidad 2 antes de estudiar números complejos y análisis de Fourier.

## Objetivos

- convertir grados a radianes y viceversa;
- interpretar seno y coseno en el círculo unitario;
- recordar signos y valores notables;
- relacionar el coseno con el ángulo entre vectores.

```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)
```

## 1. Grados y radianes

Una vuelta completa equivale a $360^\circ=2\pi$ radianes. Por tanto,

$$\theta_{\rm rad}=\theta_{\rm grados}\frac{\pi}{180},\qquad
\theta_{\rm grados}=\theta_{\rm rad}\frac{180}{\pi}. $$

```python
angulos_grados = np.array([0, 30, 45, 60, 90, 120, 180, 270, 360])
angulos_radianes = np.radians(angulos_grados)

for grados, radianes in zip(angulos_grados, angulos_radianes):
    print(f"{grados:3d} grados = {radianes:7.4f} radianes")
```

## 2. El círculo unitario

El punto del círculo unitario correspondiente al ángulo $\theta$ tiene coordenadas

$$p(\theta)=(\cos\theta,\sin\theta).$$

El coseno es la coordenada horizontal y el seno es la coordenada vertical.

```python
def dibujar_circulo_unitario(angulo_grados=45):
    theta = np.radians(angulo_grados)
    c, s = np.cos(theta), np.sin(theta)
    t = np.linspace(0, 2*np.pi, 500)

    fig, ax = plt.subplots(figsize=(6, 6))
    ax.plot(np.cos(t), np.sin(t), color="black", linewidth=1.5)
    ax.quiver(0, 0, c, s, angles="xy", scale_units="xy", scale=1,
              color="tab:purple", width=0.012)
    ax.plot([c, c], [0, s], color="tab:orange", linewidth=3, label="sin(θ)")
    ax.plot([0, c], [0, 0], color="tab:blue", linewidth=3, label="cos(θ)")
    ax.scatter([c], [s], color="tab:purple", zorder=3)
    ax.axhline(0, color="gray", linewidth=0.8)
    ax.axvline(0, color="gray", linewidth=0.8)
    ax.set_xlim(-1.25, 1.25)
    ax.set_ylim(-1.25, 1.25)
    ax.set_aspect("equal")
    ax.grid(alpha=0.25)
    ax.legend(loc="upper right")
    ax.set_title(f"θ={angulo_grados}°, cos(θ)={c:.3f}, sin(θ)={s:.3f}")
    plt.show()

dibujar_circulo_unitario(45)
```

## 3. Valores y signos por cuadrante

Antes de ejecutar, intenta anticipar el signo de cada valor.

```python
notables = np.array([0, 30, 45, 60, 90, 120, 135, 150, 180, 270, 360])
print(f"{'θ':>5} {'cos(θ)':>10} {'sin(θ)':>10}")
print("-" * 29)
for grados in notables:
    theta = np.radians(grados)
    c = np.cos(theta)
    s = np.sin(theta)
    if abs(c) < 1e-12:
        c = 0.0
    if abs(s) < 1e-12:
        s = 0.0
    print(f"{grados:5.0f} {c:10.4f} {s:10.4f}")
```

## 4. Exploración interactiva

Mueve el control y observa simultáneamente el círculo unitario y las gráficas de seno y coseno.

```python
from ipywidgets import interact, IntSlider

def explorar_trigonometria(angulo=45):
    theta = np.radians(angulo)
    c, s = np.cos(theta), np.sin(theta)
    t = np.linspace(0, 2*np.pi, 600)

    fig, axes = plt.subplots(1, 2, figsize=(12, 5))

    axes[0].plot(np.cos(t), np.sin(t), color="black")
    axes[0].quiver(0, 0, c, s, angles="xy", scale_units="xy", scale=1,
                   color="tab:purple", width=0.015)
    axes[0].plot([0, c], [0, 0], color="tab:blue", linewidth=3)
    axes[0].plot([c, c], [0, s], color="tab:orange", linewidth=3)
    axes[0].set(xlim=(-1.2, 1.2), ylim=(-1.2, 1.2),
                title=f"Círculo unitario: θ={angulo}°")
    axes[0].set_aspect("equal")
    axes[0].grid(alpha=0.25)

    grados = np.linspace(0, 360, 600)
    radianes = np.radians(grados)
    axes[1].plot(grados, np.cos(radianes), color="tab:blue", label="cos(θ)")
    axes[1].plot(grados, np.sin(radianes), color="tab:orange", label="sin(θ)")
    axes[1].scatter([angulo], [c], color="tab:blue", s=60)
    axes[1].scatter([angulo], [s], color="tab:orange", s=60)
    axes[1].axvline(angulo, color="gray", linestyle="--", linewidth=1)
    axes[1].set(xlim=(0, 360), ylim=(-1.1, 1.1), xlabel="Ángulo (grados)")
    axes[1].grid(alpha=0.25)
    axes[1].legend()

    plt.tight_layout()
    plt.show()

interact(explorar_trigonometria,
         angulo=IntSlider(min=0, max=360, step=5, value=45, description="θ"));
```

## 5. Conexión con el producto punto

Para vectores no nulos $x$ e $y$,

$$\cos\theta=\frac{x^Ty}{\|x\|\,\|y\|}. $$

El cociente siempre está en $[-1,1]$ por la desigualdad de Cauchy-Schwarz.

```python
def interpretar_angulo(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    coseno = np.clip((x @ y) / (np.linalg.norm(x) * np.linalg.norm(y)), -1, 1)
    angulo = np.degrees(np.arccos(coseno))
    if np.isclose(coseno, 0):
        tipo = "recto"
    elif coseno > 0:
        tipo = "agudo"
    else:
        tipo = "obtuso"
    return coseno, angulo, tipo

for x, y in [([1, 0], [1, 1]), ([1, 2], [-2, 1]), ([1, 0], [-1, 1])]:
    coseno, angulo, tipo = interpretar_angulo(x, y)
    print(f"x={x}, y={y}: cos(θ)={coseno:.4f}, θ={angulo:.2f}°, ángulo {tipo}")
```
